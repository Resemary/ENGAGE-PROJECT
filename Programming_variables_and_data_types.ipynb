## Introduction to Python Integrated Development Environment (IDEs)

### What is Python IDE
Python integrated development environments, also known as “Python IDEs,” are software platforms that offer programmers and developers a complete set of tools for software development in one package, specifically for the Python programming language.
A Python IDE combines a code editor, debugger, and other useful features like syntax highlighting and code auto-completion in one place. It helps programmers develop Python applications more efficiently by offering tools to write, run, and fix code within the same environment.

#### A typical Python IDE includes several important features:

- **Code Editor:** This is where you type your Python code. It often helps by using syntax highlighting (coloring keywords) and auto-complete, which suggests code as you type.

- **Interpreter:** Python uses an interpreter that reads and runs the written code line by line. This helps you test and see results quickly.

- **Debugger:** A debugger helps you find and fix bugs (errors) in your code. You can go through your code step-by-step to understand what's going wrong.

- **Console/Terminal:** This is where you see the output of your code or type commands directly to interact with Python.

- **Project Tools:** If you're working on a big project, the IDE helps you organize your files and manage the code better.

**Variables and Data types**

Variables in Python are essentially named references pointing to objects in memory or we can say, a variable is like a container that holds information. You can give it a name and store a value inside it using the = sign

Variables help you keep and use information in your program, like a name, number, or a yes/no answer.

You need to create (declare) a variable before using it in your code.




Here are some examples:

```python
name = "Alice"       # This is a string (text)
age = 25             # This is an integer (whole number)
height = 5.6         # This is a float (decimal number)
is_student = True    # This is a boolean (True or False)
```

**Operators**

- Operators are symbols used to perform operations on data.  Here are some essential operators:

- Arithmetic operators (+, -, *, /) perform calculations on numerical data.

- Comparison operators (==, !=, <, >, <=, >=) compare values and return True or False based on the condition.

- Logical operators (and, or, not) combine conditions to create more complex expressions.

**Arithmetic Operators**

- Addition (+): Adds two operands (numbers).

```python
x = 5 + 3
print(x)
```

- Subtraction (-): Subtracts the second operand from the first.

```python
y = 12 - 2
print(y)
```

- Multiplication (*): Multiplies two operands.

```python
z = 4 * 6
print(z)
```

- Division (/): Divides the first operand by the second (results in a float by default).

```python
a = 12 / 3
print(a)
```

- Integer Division (//): Performs integer division, resulting in the whole number quotient.

```python
b = 12 // 3
print(b)
```

- Modulo (%): Gives the remainder after division.

```python
c = 10 % 3
print(c)
```

- Exponentiation (**): Raises the first operand to the power of the second.

```python
d = 2 ** 3
print(d)
```

Comparison Operators

These operators compare values and return Boolean results (True or False).

- Equal (==): Checks if two operands are equal.

```python
7 == 7
```

- Not equal (!=): Checks if two operands are not equal.

```python
f = 5 != 8
print(f)
```

- Greater than (>): Checks if the left operand is greater than the right.

```python
g = 15 > 10
print(g)
```

- Less than (<): Checks if the left operand is less than the right.

```python
h = 3 < 1
print(h)
```

- Greater than or equal to (>=): Checks if the left operand is greater than or equal to the right.

```python
i = 9 >= 9
print(i)
```

- Less than or equal to (<=): Checks if the left operand is less than or equal to the right.

```python
j = 2 <= 4
print(j)
```

**Logical Operators**

These operators combine conditional statements.

- AND (and): Returns True only if both operands are True.

```python
k = (7 > 5) and (3 < 6)
print(k)
```

- OR (or): Returns True if at least one operand is True.

```python
l = (8 != 8) or (12 > 10)
print(l)
```

- NOT (not): Inverts the truth value of a single operand.

```python
m = not (9 == 9)
print(m)
```

**Assignment Operators**

These operators assign values to variables.

- Simple assignment (=): Assigns the value on the right to the variable on the left.

```python
n = 20
print(n)
```

**Dictionary**

- A dictionary is a collection of key-value pairs. It allows you to store and retrieve data based on keys rather than indexes.

- Dictionaries can be created using curly braces {} or the dict() function. Each key-value pair is separated by a colon :.

```python
# using curly braces
my_dict = {'name': 'John', 'age': 30, 'city': 'Nairobi'}
print(my_dict)

# using the dict() function
another_dict = dict(name='Alice', age=25, city='Kisumu')
print(another_dict)
```

- You can access the value associated with a key using square brackets [].
- If the key doesn't exist, it will raise a KeyError. To avoid this, you can use the get() method.

```python
name = my_dict['name']  # Accessing value using key
my_dict['age'] = 35  # Modifying value associated with key
```

Dictionary Methods

- keys(): Returns a view object that displays a list of all the keys in the dictionary.
- values(): Returns a view object that displays a list of all the values in the dictionary.
- items(): Returns a view object that displays a list of key-value pairs in the dictionary

```python
print(my_dict.keys())
print(my_dict.values())
print(my_dict.items())
```

**Set**

In Python, a set is an unordered collection of unique elements. This means that each element in a set is unique, and the order of elements doesn't matter.

Creating a Set:
Sets can be created using curly braces {} or the set() function. For example:

```python
# Creating a set using curly braces
my_set = {1, 2, 3, 4, 5}

# Creating a set using the set() function
another_set = set([4, 5, 6, 7, 8])
```

You can add elements to a set using the add() method, and remove elements using the remove() or discard() methods.

```python
my_set.add(6)
my_set.discard(4)
print(my_set)
```

Operations on Sets

- Union: Combines two sets, returning a new set containing all unique elements from both sets.
- Intersection: Returns a new set containing only the elements that are common to both sets.
- Difference: Returns a new set containing elements that are in the first set but not in the second set.
- Subset and Superset: Checks if one set is a subset or superset of another.

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

union_set = set1.union(set2)
print(union_set)
```

```python
intersection_set = set1.intersection(set2)
print(intersection_set)
```

```python
difference_set = set1.difference(set2)
print(difference_set)
```

## **Conditions and loops**

Conditions allow you to execute different blocks of code based on whether certain conditions are true or false.

Conditional statements include if, else, and elif. The basic idea in pseudocode is:

if (some condition is true) then do X otherwise do Y

Here, indenting and whitespace are very important. After the if (some condition) part, the do X part must be indented by exactly 1 tab; otherwise, you'll get an error.

a) if Statement

- The if statement checks a condition and executes a block of code if the condition is true.

```python
x = 10

if x > 0:
    print("x is positive")
```

b) elif Statement

The elif statement allows you to check multiple conditions if the previous conditions are not met.

```python
x = 0

if x > 0:
    print("x is positive")
elif x == 0:
    print("x is zero")
```

c) else Statement

The else statement executes a block of code if none of the previous conditions are true.

```python
x = -5

if x > 0:
    print("x is positive")
else:
    print("x is not positive")
```

For more info and details on the if,elif and else statements.The [python documentation](https://docs.python.org/3/reference/compound_stmts.html#if) is a good place to start.

### **Loops**

Loops are used to execute a block of code repeatedly. In Python, you have the for loop and the while loop.

a) for Loop

A for loop is used to iterate over a sequence (such as a list, string among others).

```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

b) while Loop

```python
i = 0

while i < 5:
    print(i)
    i += 1
```

##**Python Functions and Procedures**

In Python, functions and procedures are both blocks of reusable code, but they differ slightly in their purpose and behavior

**Functions** is a block of code that performs a specific task and returns a value.
Functions are defined using the `def` keyword, followed by a name, parentheses (which may include parameters), and a colon.

You can pass data, known as parameters, into a function.

A function can return data as a result.

**Procedures**

A procedure is like a function but it does not return a result. It just performs an action like printing or saving something.

It takes input but does not return a value.
It simply performs a task


**Creating a Function**

In Python a function is defined using the ***def***  keyword:

```python
def my_function():
  print("Hello from a function")
```

**Calling a Function**

To call a function, use the function name followed by parenthesis:

```python
def my_function():
  print("Hello from a function")

my_function()
```

**Function with Parameters**

Parameters allow you to pass information into a function.

**Arguments**

Information can be passed into functions as arguments.

Arguments are specified after the function name, inside the parentheses. You can add as many arguments as you want, just separate them with a comma.

The following example has a function with one argument (fname). When the function is called, we pass along a first name, which is used inside the function to print the full name:

```python
def my_function(fname):
  print(fname + " Refsnes")

my_function("Emil")
my_function("Tobias")
my_function("Linus")
```

##Introduction to Python Libraries

A Python library is a set of ready-made tools (code) that helps you do things faster and easier in your program. Instead of writing everything yourself, you just use a library.


**Why Use Libraries?**

- They save time

- They make your code shorter and easier to read

- They help you do big tasks with just a few lines of code



Each Python library has its own purpose and helps with specific tasks.

**Common Python Libraries**

1. pandas: Helps you work with data in tables, like rows and columns. It’s great for reading, cleaning, and analyzing data (like spreadsheets).

2. Numpy: Makes working with numbers and arrays (lists of numbers) easier and faster. It’s often used for math and scientific calculations

3. matplotlib: Used to make charts and graphs. You can create bar graphs, line charts, and more to help you visualize your data.

4. SciPy: Builds on NumPy and helps with more advanced math, like solving equations, working with signals, or doing statistics.

5. Scikit-Learn: Used for machine learning. It helps you build models that can make predictions or classify data (like predicting prices or sorting emails).

6. PyTorch: A powerful tool for deep learning and AI. It helps you build and train smart computer programs that can learn from data (like image recognition).



**How to Use a Library**
Before using a library in Python, you need to import it using the ***import*** keyword.

You can also give it a short name, called an alias, to make it easier to use.

For example:


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```
