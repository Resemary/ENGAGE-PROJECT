Hypothesis testing :

Example of comparing blood pressure between group A nd B

```python
import numpy as np
from scipy import stats
# Given summary statistics for two samples
mean1 =120   # Mean of sample 1
std1 = 10     # Standard deviation of sample 1
n1 = 30        # Sample size of group 1

mean2 = 130   # Mean of sample 2
std2 = 12    # Standard deviation of sample 2
n2 = 30        # Sample size of group 2

# Perform independent two-sample t-test (assuming unequal variances by default)
t_stat, p_value = stats.ttest_ind_from_stats(
    mean1=mean1, std1=std1, nobs1=n1,
    mean2=mean2, std2=std2, nobs2=n2,
    equal_var=True  # Welch's t-test (does not assume equal variances)
)
p_value_left_tailed = p_value / 2
# Output results
print(f"t-statistic: {t_stat:.4f}")
print(f"p-value: {p_value_left_tailed:.4f}")
```

Illustration in Python:
Is there a significant difference in BMI values between smokers and non smokers

```python

import pandas as pd
from scipy import stats
df=pd.read_csv("/content/Diabetes_data.xls")
group_one = df[df['smoking'] == 'Nonsmoker']['BMI']
group_two = df[df['smoking'] == 'Smoker']['BMI']
t_statistic, p_value = stats.ttest_ind(group_one, group_two)
print(f"T-Statistic: {t_statistic}, p-Value: {p_value}")
```

Demonstration for two sample z-test:
compare children who had diarrhea in rural area to those who had diarrhea in urban area

we require the sample sizes of groups(rural and urban) and number of those who had diarrhea in each group

to get this we obtain two way contingency table

```python
ch_dia=pd.read_csv("/content/childhealthdata (6).csv")
ctable = pd.crosstab(
    ch_dia[' diarrhea_status'],
    ch_dia['residence'],
    margins=True,
    margins_name="Total"
)

print(ctable)
```

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

# Sample data: number of successes and sample sizes
dia_yes = np.array([46, 48])  # Number of those who diarrhea in each group
samsize = np.array([452, 346])   # sample size of  each group

# Perform two-sample z-test for proportions
# Alternative options: 'two-sided' (default), 'larger', 'smaller'
z_stat, p_value = proportions_ztest(dia_yes, samsize, alternative='two-sided')

# Output results
print(f"z-statistic: {z_stat:.4f}")
print(f"p-value: {p_value:.4f}")
```

```python
import scipy.stats as stats
import statsmodels.api as sm

#create contingency table
table = pd.crosstab(df['smoking'], df['diabetes'])

# Convert to NumPy array for analysis
tab = table.to_numpy()

# Fisher’s Exact Test for Odds Ratio
odds_ratio, p_value = stats.fisher_exact(tab)
print(f"Odds Ratio: {odds_ratio:.2f}")
print(f"p-value: {p_value:.4f}")

# Confidence interval for odds ratio using statsmodels
table2 = sm.stats.Table2x2(tab)
ci_low, ci_high = table2.oddsratio_confint()
print(f"95% CI for Odds Ratio: ({ci_low:.2f}, {ci_high:.2f})")
```
